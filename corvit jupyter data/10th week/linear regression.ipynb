```python
import numpy as np

# 1. Dataset from your screen
x_train = np.array([500, 800, 1000, 1200, 1500, 1800, 2000])
y_train = np.array([100, 150, 180, 210, 260, 300, 330])

def compute_cost(x,y,w,b):
    m=len(x)
    cost= 0
    for i in range(m):
        f_wb= w*x[i]+b
        cost +=(f_wb-y[i])**2
        return (1/(2*m))*cost
# 2. Trial-and-error parameters from Part 1
b = 0  # Initial value set as 0
w_values = np.arange(0.10, 0.21, 0.01) # 0.10 to 0.20 with 0.01 increment
print(f"{'w value':<10} | {'Cost J(w,b)':<15}")
print("-" * 30)
best_w = 0
min_cost = float('inf')
for w in w_values:
    current_cost =compute_cost(x_train,y_train,w,b)

    print(f"{w:.2f}       | {current_cost:.2f}")
    
    if current_cost < min_cost:
        min_cost = current_cost
        best_w = w
        
print("-" * 30)
print(f"Best w found: {best_w:.2f} with Cost: {min_cost:.2f}")
```
