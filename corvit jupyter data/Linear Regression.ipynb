```python
area = [0, 1, 2, 3, 4, 5]
price = [0, 2, 4, 6, 8, 10]

w = 0
b = 1

samples = len(area)

x = area
y = price

total_sse = 0

for i in range(samples):
    f_x = w * x[i] + b
    error = y[i] - f_x
    sse = error ** 2
    total_sse += sse
    
    print(f"Actual Price Y = {y[i]}, Predicted = {f_x}, Error = {error}, Squared Error = {sse}")

MSE = total_sse / samples
print(f"MSE = {MSE}")
```
