```python
import numpy as np

# 1. Dataset from the table
x = np.array([500, 800, 1000, 1200, 1500, 1800, 2000])
y = np.array([100, 150, 180, 210, 260, 300, 330])

# 2. Define trial ranges
w_values = np.arange(0.10, 0.21, 0.01) # 0.10 to 0.20
b_values = np.arange(0, 36, 5)         # 0 to 35

# INITIALIZE VARIABLES (This fixes your NameError)
best_w = 0
best_b = 0
min_mse = float('inf')  # Start with infinity so any first MSE is lower

# 3. Trial-and-Error Loop
for w in w_values:
    for b in b_values:
        y_pred = w * x + b
        mse = np.mean((y - y_pred)**2)
        
        if mse < min_mse:
            min_mse = mse
            best_w = w
            best_b = b

print(f"Best w: {best_w:.2f}")
print(f"Best b: {best_b}")
print(f"Lowest MSE: {min_mse:.2f}")

# 4. Part 3: Predictions
test_areas = np.array([700, 900, 1300, 1700, 2100])
print("\nPart 3 Predictions:")
for area in test_areas:
    prediction = best_w * area + best_b
    print(f"Area: {area} sqft -> Predicted Price: ${prediction:.2f}k")
```
