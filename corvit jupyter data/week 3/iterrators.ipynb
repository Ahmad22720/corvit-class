## iterators in lists

```python
a=[1,2,3,4,5]
itr_a=iter(a)
print(next(itr_a))
print(next(itr_a))
print(next(itr_a))
print(next(itr_a))
print(next(itr_a))
```

## iterators in tuple

```python
b=('apple','banana', 'cherry','orange')
for x in b:
    print(x)
```

## string

```python
c='apple'
myit=iter(c)
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

## dictionaries

```python
t = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
c=iter(t)
print(next(c))
print(next(c))
print(next(c))

for i in t:
 value=t[i]   
 print (value)
```
