## map function

```python
sqr = lambda x:x**2
a= [2,3,4,5,6,7,8,99,9]
swreq=map(sqr,a)
print(list(swreq))
```

# filter function

```python
b=[2,4,7,8,90,7,688,5,7,8]
def iseven(x):
 return x%2==0
even_number=filter(iseven,b)
print('even numbers are :',list(even_number))
def isodd (x):
 return x%2!=0
odd_num=filter(isodd,b)
print('odd numbers are :',list(odd_num))
```
