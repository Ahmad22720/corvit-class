```python
a=[8,77,99,33,22,44]
print(a)
```

```python
# created set as list
list_a=list (a)
# run loop from end
rev=list_a[::-1]
print(rev)
```

# find unique values

```python
customer=['mango','apple','dates',
         'pineapple','apple','mango',
         'dates','apple','mango']
print(customer)
```

```python
value=list(set(customer))
print(value)
```

```python
for order in customer:
    print("the order is :", order)
```

```python
count=len(customer)
print('the orders are :', count)
```
